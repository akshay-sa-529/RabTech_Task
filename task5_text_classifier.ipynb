# Task 5 — Deep Learning Text Classifier (PyTorch)


```python
# Step 1–3: Imports, dataset, vectorization
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

texts = ["I love AI", "I hate bugs", "Deep learning is amazing", "This code fails"]
labels = [1, 0, 1, 0]


vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(texts).toarray()
y = torch.tensor(labels, dtype=torch.float32).unsqueeze(1)

```

```python
# Step 4–5: Split data + define model
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
train_ds = TensorDataset(torch.tensor(X_train, dtype=torch.float32), y_train.unsqueeze(1))
train_dl = DataLoader(train_ds, batch_size=2, shuffle=True)


class TextClassifier(nn.Module):
    def __init__(self, input_dim):
        super(TextClassifier, self).__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, 16),
            nn.BatchNorm1d(16),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(16, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        return self.net(x)

model = TextClassifier(X.shape[1])
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)
```

```python
epochs = 20
losses = []
for epoch in range(epochs):
    epoch_loss = 0
    for xb, yb in train_dl:
        optimizer.zero_grad()
        preds = model(xb)
        loss = criterion(preds, yb)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()
    losses.append(epoch_loss)
    print(f"Epoch {epoch+1}/{epochs}, Loss: {epoch_loss:.4f}")
```

```python
# Step 7: Plot training loss
plt.plot(losses)
plt.title("Training Loss Curve")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()
```
